```python
# ============================================================
# AmhMedQA Dataset Preprocessing code
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# 1. Load the dataset
# ------------------------------------------------------------

file_path = "AmhMedQA.csv"

df = pd.read_csv(
    file_path,
    encoding="utf-8"
)
print("Dataset shape:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())


# ------------------------------------------------------------
# 2. Check the required columns
# ------------------------------------------------------------

required_columns = ["Question", "Answer", "Category"]

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing columns: {missing_columns}"
    )

# Keep only the three required columns
df = df[required_columns].copy()


# ------------------------------------------------------------
# 3. Clean missing values and whitespace
# ------------------------------------------------------------

df = df.dropna(
    subset=["Question", "Answer", "Category"]
)

df["question"] = (
    df["Question"]
    .astype(str)
    .str.strip()
)

df["answer"] = (
    df["Answer"]
    .astype(str)
    .str.strip()
)

df["category"] = (
    df["Category"]
    .astype(str)
    .str.strip()
)


# ------------------------------------------------------------
# 4. Display final dataset information
# ------------------------------------------------------------

print("\nFinal dataset shape:")
print(df.shape)

print("\nFirst five records:")
print(df.head())


# ============================================================
# 5. QUESTION-ANSWER CATEGORY DISTRIBUTION
# ============================================================

distribution = (
    df["category"]
    .value_counts()
    .sort_index()
    .reset_index()
)

distribution.columns = [
    "Category",
    "Number_of_Records"
]

# Calculate percentage
distribution["Percentage"] = (
    distribution["Number_of_Records"]
    / len(df)
    * 100
).round(2)


# ------------------------------------------------------------
# 6. Add total row
# ------------------------------------------------------------

total_row = pd.DataFrame({
    "Category": ["Total"],
    "Number_of_Records": [
        distribution["Number_of_Records"].sum()
    ],
    "Percentage": [
        distribution["Percentage"].sum()
    ]
})

distribution_with_total = pd.concat(
    [distribution, total_row],
    ignore_index=True
)


# ------------------------------------------------------------
# 7. Display distribution table
# ------------------------------------------------------------

print("\n========================================")
print("QUESTION-ANSWER CATEGORY DISTRIBUTION")
print("========================================")

print(distribution_with_total.to_string(index=False))


# ============================================================
# 8.======= SAVE DISTRIBUTION TABLE
# ============================================================

distribution_csv = (
    "/content/AmhMedQA_category_distribution.csv"
)

distribution_with_total.to_csv(
    distribution_csv,
    index=False,
    encoding="utf-8"
)

print(
    f"Distribution table saved to:\n{distribution_csv}"
)


# ============================================================
# 10. SAVE DISTRIBUTION TABLE AS EXCEL
# ============================================================

distribution_excel = (
    "/content/AmhMedQA_category_distribution.xlsx"
)

distribution_with_total.to_excel(
    distribution_excel,
    index=False
)

print(
    f"Distribution Excel file saved to:\n{distribution_excel}"
)


# ============================================================
# 11. CREATE CATEGORY DISTRIBUTION BAR GRAPH
# ============================================================

plt.figure(figsize=(8, 5))

bars = plt.bar(
    distribution["Category"],
    distribution["Number_of_Records"]
)

plt.xlabel("Question Category")
plt.ylabel("Number of Records")

plt.title(
    "Distribution of Question-Answer Categories in AmhMedQA"
)

# Display count above each bar
for bar in bars:

    height = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        str(int(height)),
        ha="center",
        va="bottom"
    )

plt.tight_layout()


# ============================================================
# 12. SAVE GRAPH
# ============================================================

graph_file = (
    "/content/AmhMedQA_category_distribution.png"
)

plt.savefig(
    graph_file,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print(
    f"Category distribution graph saved to:\n{graph_file}"
)


# ============================================================
# 13. FINAL VERIFICATION
# ============================================================

print("\n========================================")
print("FINAL VERIFICATION")
print("========================================")

print(
    "Number of QA records:",
    len(df)
)

print(
    "Number of columns:",
    len(df.columns)
)

print(
    "Columns:",
    df.columns.tolist()
)

print("\nCategory distribution:")
print(distribution)
```
